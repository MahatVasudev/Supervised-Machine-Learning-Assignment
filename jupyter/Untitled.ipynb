```python
import torch
from models.CONVLSTM import ConvLSTM2Layers
from sklearn.metrics import mean_absolute_error, r2_score
import numpy as np 

device = "cuda" if torch.cuda.is_available() else "cpu"
model_2_layer = ConvLSTM2Layers().to(device) 
model_2_layer.load_state_dict(torch.load("saved_models/ConvLSTM2Layers_25_checkpoint.pt", map_location=device)["model"])
model_2_layer.eval()

y_trues = []
y_preds = []
with torch.no_grad(): 
    for X, Y in test_loader:
        X, Y = X.to(device), Y.to(device) 
        Y_pred = model_2_layer(X) 

        y_trues.append(Y.cpu().numpy().flatten())
        y_preds.append(Y_pred.cpu().numpy().flatten())

y_trues = np.concatenate(y_trues) 
y_preds = np.concatenate(y_preds)

mae = mean_absolute_error(y_trues, y_preds)
r2 = r2_score(y_trues, y_preds)
```

```python
from ML_Model_Building.models.CONVLSTM import ConvLSTM
import torch 

device = "cuda" if torch.cuda.is_available() else "cpu"
model_1_layer = ConvLSTM().to(device) 
model_1_layer.load_state_dict(torch.load("ML_Model_Building/saved_models/ConvLSTM_25_checkpoint.pt", map_location=device)["model"])
model_1_layer.eval()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# ---------------------------
# 1. Model Inference (multi-day prediction)
# ---------------------------
model_1_layer.eval()

# Get one batch of data
X_batch, Y_batch = next(iter(test_loader))
X_batch, Y_batch = X_batch.to(device), Y_batch.to(device)

# Select one sample (keep batch dim)
X_input = X_batch[0:1]  # shape (1, seq_len, 1, H, W)
num_future_days = 7  # how many future predictions you want

predictions = []

with torch.no_grad():
    for day in range(num_future_days):
        # Predict next day
        Y_pred = model_1_layer(X_input)
        predictions.append(Y_pred.cpu().numpy())

        # Append prediction as next input (auto-regressive)
        X_input = torch.cat([X_input[:, 1:], Y_pred.unsqueeze(1)], dim=1)

# Stack predictions
Y_pred_sample = np.concatenate(predictions, axis=0)  # shape: (num_future_days, 1, H, W)
print("Predicted future frames shape:", Y_pred_sample.shape)

# ---------------------------
# 2. Map & Animation Setup
# ---------------------------

# Define geographic extents (ensure you set these correctly)
min_lon, max_lon = -180.0, 180.0
min_lat, max_lat = -90.0, 90.0


# Create latitude & longitude grids
lats = np.linspace(min_lat, max_lat, Y_pred_sample.shape[2])
lons = np.linspace(min_lon, max_lon, Y_pred_sample.shape[3])
lon_grid, lat_grid = np.meshgrid(lons, lats)

# Create figure
fig = plt.figure(figsize=(16, 6))
ax = plt.axes(projection=ccrs.PlateCarree())

# Add map features
ax.coastlines(resolution='10m', color='gray', linewidth=0.5)
ax.add_feature(cfeature.BORDERS, linestyle=':', linewidth=0.5)
ax.set_extent([min_lon, max_lon, min_lat, max_lat], crs=ccrs.PlateCarree())

# Initial frame
pcm = ax.pcolormesh(
    lon_grid, lat_grid, Y_pred_sample[0, 0],
    cmap='hot', transform=ccrs.PlateCarree(),
    shading='auto', alpha=0.7
)
plt.colorbar(pcm, ax=ax, label='Predicted Fire Probability')
title = ax.set_title("Day 1 Prediction")

# ---------------------------
# 3. Animation Function
# ---------------------------
def update(frame):
    pcm.set_array(Y_pred_sample[frame, 0].ravel())  # update heatmap
    title.set_text(f"Day {frame + 1} Prediction")
    return [pcm, title]

# ---------------------------
# 4. Animate
# ---------------------------
anim = FuncAnimation(fig, update, frames=Y_pred_sample.shape[0], interval=500, blit=False)

# Display inline in Jupyter
HTML(anim.to_jshtml())

```
